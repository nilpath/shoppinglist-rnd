```python
import os

import dspy
import dotenv

from shoppinglist_rnd.agent import ShoppingListAgent
from shoppinglist_rnd.models import Ingredient
```

```python
PROJECT_ROOT = os.path.abspath(os.path.join(os.path.pardir, os.path.pardir, os.path.pardir))
DOTENV_FILE = os.path.join(PROJECT_ROOT, '.env')
dotenv.load_dotenv(DOTENV_FILE)
```

```python
lm = dspy.LM("openai/gpt-4o-mini")
dspy.configure(lm=lm)
```

```python
ingredients = [
    Ingredient(name="Apple", quantity=2, unit="pcs"),
    Ingredient(name="Milk", quantity=1, unit="liter"),
    Ingredient(name="Chicken Breast", quantity=500, unit="grams"),
    Ingredient(name="Frozen Peas", quantity=300, unit="grams"),
]
```

```python
agent = ShoppingListAgent()
result = agent(ingredients=ingredients)
```
